```python

# ============================================================
# ZEROQ AI - NOTEBOOK 04
# INDIVIDUAL DATASET EXPLORATORY DATA ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
import warnings
import re

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 100

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")


# ------------------------------------------------------------
# 2. DEFINE DATASET FILENAMES
# ------------------------------------------------------------

DATASET_FILES = {
    "products": "products_cleaned.csv",
    "customers": "customers_cleaned.csv",
    "facilities": "facilities_cleaned.csv",
    "production_lines": "production_lines_cleaned.csv",
    "materials": "materials_cleaned.csv",
    "production_batches": "production_batches_cleaned.csv",
    "complaints": "complaints_cleaned.csv",
    "quality_inspections": "quality_inspections_cleaned.csv",
    "investigations": "investigations_cleaned.csv",
    "corrective_actions": "corrective_actions_cleaned.csv",
    "data_dictionary": "data_dictionary_cleaned.csv"
}

REQUIRED_FILES = set(DATASET_FILES.values())


# ------------------------------------------------------------
# 3. AUTOMATICALLY LOCATE PROCESSED DATA FOLDER
# ------------------------------------------------------------

print("\nSearching for processed datasets...")

search_roots = [Path.cwd(), *Path.cwd().parents]

PROCESSED_DATA_PATH = None

for root in search_roots:

    for file_path in root.rglob("products_cleaned.csv"):

        candidate_folder = file_path.parent

        available_files = {
            file.name for file in candidate_folder.glob("*.csv")
        }

        if REQUIRED_FILES.issubset(available_files):
            PROCESSED_DATA_PATH = candidate_folder
            break

    if PROCESSED_DATA_PATH is not None:
        break

if PROCESSED_DATA_PATH is None:

    raise FileNotFoundError(
        "Could not locate all 11 cleaned datasets. "
        "Execute Notebook 2 before proceeding."
    )

PROJECT_PATH = PROCESSED_DATA_PATH.parent

PREPROCESSING_PATH = PROJECT_PATH / "preprocessing"
VISUALIZATION_PATH = PROJECT_PATH / "visualizations"

INDIVIDUAL_EDA_PATH = VISUALIZATION_PATH / "individual_eda"

REPORT_PATH = PREPROCESSING_PATH / "reports"

INDIVIDUAL_EDA_PATH.mkdir(parents=True, exist_ok=True)
REPORT_PATH.mkdir(parents=True, exist_ok=True)

print("\nPROJECT CONFIGURATION")
print("-" * 60)
print("Project:", PROJECT_PATH)
print("Processed data:", PROCESSED_DATA_PATH)
print("Visualizations:", INDIVIDUAL_EDA_PATH)


# ------------------------------------------------------------
# 4. LOAD ALL CLEANED DATASETS
# ------------------------------------------------------------

datasets = {}

for dataset_name, filename in DATASET_FILES.items():

    file_path = PROCESSED_DATA_PATH / filename

    if not file_path.exists():
        raise FileNotFoundError(f"Missing dataset: {file_path}")

    df = pd.read_csv(file_path)

    datasets[dataset_name] = df

    print(
        f"{dataset_name:25} | "
        f"Shape: {df.shape}"
    )

print("\nTotal datasets loaded:", len(datasets))


# ------------------------------------------------------------
# 5. CREATE OUTPUT DIRECTORIES FOR EACH DATASET
# ------------------------------------------------------------

for dataset_name in datasets:

    folder = INDIVIDUAL_EDA_PATH / dataset_name

    folder.mkdir(parents=True, exist_ok=True)

print("\nEDA output folders created.")


# ------------------------------------------------------------
# 6. HELPER FUNCTIONS
# ------------------------------------------------------------

def clean_filename(value):

    value = re.sub(r"[^a-zA-Z0-9_-]+", "_", str(value))

    return value.strip("_").lower()


def save_plot(dataset_name, plot_name):

    output_folder = INDIVIDUAL_EDA_PATH / dataset_name

    output_file = (
        output_folder / f"{clean_filename(plot_name)}.png"
    )

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=150,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()

    print("Saved:", output_file.name)


def get_categorical_columns(df):

    return list(
        df.select_dtypes(
            include=["object", "category", "string"]
        ).columns
    )


def get_numeric_columns(df):

    numeric_columns = list(
        df.select_dtypes(include=np.number).columns
    )

    # Exclude likely identifier columns from numerical EDA
    return [
        column for column in numeric_columns
        if not column.lower().endswith("_id")
        and column.lower() not in ["id", "year", "month"]
    ]


def get_date_columns(df):

    return [
        column for column in df.columns
        if (
            "date" in column.lower()
            or "timestamp" in column.lower()
            or column.lower().endswith("_at")
        )
    ]


def plot_categorical_distribution(df, dataset_name, column):

    counts = (
        df[column]
        .fillna("Missing")
        .astype(str)
        .value_counts()
        .head(15)
    )

    if counts.empty:
        return

    plt.figure(figsize=(11, 6))

    sns.barplot(
        x=counts.values,
        y=counts.index
    )

    plt.title(
        f"{dataset_name.title()} - {column} Distribution"
    )

    plt.xlabel("Number of Records")
    plt.ylabel(column.replace("_", " ").title())

    save_plot(
        dataset_name,
        f"{column}_distribution"
    )


def plot_numeric_distribution(df, dataset_name, column):

    values = pd.to_numeric(
        df[column],
        errors="coerce"
    ).dropna()

    if len(values) < 2:
        return

    plt.figure(figsize=(10, 6))

    sns.histplot(
        values,
        kde=True,
        bins=30
    )

    plt.title(
        f"{dataset_name.title()} - {column} Distribution"
    )

    plt.xlabel(column.replace("_", " ").title())
    plt.ylabel("Frequency")

    save_plot(
        dataset_name,
        f"{column}_distribution"
    )


def plot_numeric_boxplot(df, dataset_name, column):

    values = pd.to_numeric(
        df[column],
        errors="coerce"
    ).dropna()

    if len(values) < 2:
        return

    plt.figure(figsize=(10, 4))

    sns.boxplot(x=values)

    plt.title(
        f"{dataset_name.title()} - {column} Boxplot"
    )

    plt.xlabel(column.replace("_", " ").title())

    save_plot(
        dataset_name,
        f"{column}_boxplot"
    )


def plot_date_trend(df, dataset_name, column):

    dates = pd.to_datetime(
        df[column],
        errors="coerce"
    )

    valid_dates = dates.dropna()

    if len(valid_dates) < 2:
        return

    yearly_counts = (
        valid_dates.dt.year
        .value_counts()
        .sort_index()
    )

    if len(yearly_counts) < 2:
        return

    plt.figure(figsize=(12, 5))

    sns.lineplot(
        x=yearly_counts.index,
        y=yearly_counts.values,
        marker="o"
    )

    plt.title(
        f"{dataset_name.title()} - Yearly Record Trend"
    )

    plt.xlabel("Year")
    plt.ylabel("Number of Records")

    plt.xticks(rotation=45)

    save_plot(
        dataset_name,
        f"{column}_yearly_trend"
    )


# ------------------------------------------------------------
# 7. GENERATE DATASET STATISTICAL SUMMARIES
# ------------------------------------------------------------

dataset_summaries = []
column_summaries = []

for dataset_name, df in datasets.items():

    print("\n" + "=" * 75)
    print("DATASET PROFILE:", dataset_name.upper())
    print("=" * 75)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Memory (MB):", round(
        df.memory_usage(deep=True).sum() / (1024 ** 2), 3
    ))

    print("\nData types:")
    print(df.dtypes)

    print("\nStatistical summary:")
    display(df.describe(include="all").T)

    dataset_summaries.append({

        "Dataset": dataset_name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Missing_Values": int(df.isna().sum().sum()),
        "Duplicate_Rows": int(df.duplicated().sum()),
        "Numeric_Columns": len(get_numeric_columns(df)),
        "Categorical_Columns": len(get_categorical_columns(df)),
        "Date_Columns": len(get_date_columns(df))

    })

    for column in df.columns:

        column_summaries.append({

            "Dataset": dataset_name,
            "Column": column,
            "Data_Type": str(df[column].dtype),
            "Unique_Values": int(df[column].nunique(dropna=True)),
            "Missing_Values": int(df[column].isna().sum()),
            "Missing_Percentage": round(
                df[column].isna().mean() * 100, 2
            )

        })


dataset_summary_df = pd.DataFrame(dataset_summaries)

column_summary_df = pd.DataFrame(column_summaries)

print("\nOVERALL DATASET SUMMARY")
display(dataset_summary_df)


# ------------------------------------------------------------
# 8. INDIVIDUAL DATASET VISUALIZATIONS
# ------------------------------------------------------------

for dataset_name, df in datasets.items():

    print("\n" + "=" * 80)
    print("GENERATING EDA:", dataset_name.upper())
    print("=" * 80)

    categorical_columns = get_categorical_columns(df)
    numeric_columns = get_numeric_columns(df)
    date_columns = get_date_columns(df)

    # --------------------------------------------------------
    # 8.1 CATEGORICAL DISTRIBUTIONS
    # --------------------------------------------------------

    for column in categorical_columns:

        # Avoid plotting high-cardinality columns such as IDs
        if df[column].nunique(dropna=True) > 100:
            continue

        plot_categorical_distribution(
            df,
            dataset_name,
            column
        )

    # --------------------------------------------------------
    # 8.2 NUMERICAL DISTRIBUTIONS AND BOXPLOTS
    # --------------------------------------------------------

    for column in numeric_columns:

        if df[column].nunique(dropna=True) < 2:
            continue

        plot_numeric_distribution(
            df,
            dataset_name,
            column
        )

        plot_numeric_boxplot(
            df,
            dataset_name,
            column
        )

    # --------------------------------------------------------
    # 8.3 DATE-BASED TRENDS
    # --------------------------------------------------------

    for column in date_columns:

        plot_date_trend(
            df,
            dataset_name,
            column
        )

    # --------------------------------------------------------
    # 8.4 NUMERICAL CORRELATION HEATMAP
    # --------------------------------------------------------

    correlation_columns = [
        column for column in numeric_columns
        if df[column].nunique(dropna=True) > 1
    ]

    if len(correlation_columns) >= 2:

        correlation_matrix = df[
            correlation_columns
        ].corr()

        plt.figure(figsize=(12, 8))

        sns.heatmap(
            correlation_matrix,
            annot=True,
            fmt=".2f",
            cmap="coolwarm",
            center=0
        )

        plt.title(
            f"{dataset_name.title()} - Correlation Heatmap"
        )

        save_plot(
            dataset_name,
            "correlation_heatmap"
        )

    print("EDA completed for:", dataset_name)


# ------------------------------------------------------------
# 9. SPECIALIZED COMPLAINT ANALYSIS
# ------------------------------------------------------------

complaints_df = datasets["complaints"]

print("\n" + "=" * 80)
print("SPECIALIZED COMPLAINT ANALYSIS")
print("=" * 80)

complaint_category_columns = [
    column for column in complaints_df.columns
    if any(
        keyword in column.lower()
        for keyword in ["category", "type", "subcategory"]
    )
]

severity_columns = [
    column for column in complaints_df.columns
    if "severity" in column.lower()
]

for column in complaint_category_columns + severity_columns:

    if column not in complaints_df.columns:
        continue

    counts = (
        complaints_df[column]
        .fillna("Missing")
        .astype(str)
        .value_counts()
        .head(15)
    )

    if counts.empty:
        continue

    plt.figure(figsize=(11, 6))

    sns.barplot(
        x=counts.values,
        y=counts.index
    )

    plt.title(f"Complaint Analysis - {column}")

    plt.xlabel("Number of Complaints")
    plt.ylabel(column.replace("_", " ").title())

    save_plot(
        "complaints",
        f"special_{column}"
    )


# ------------------------------------------------------------
# 10. SPECIALIZED PRODUCTION BATCH ANALYSIS
# ------------------------------------------------------------

batches_df = datasets["production_batches"]

print("\n" + "=" * 80)
print("SPECIALIZED PRODUCTION BATCH ANALYSIS")
print("=" * 80)

process_keywords = [
    "temperature",
    "humidity",
    "pressure",
    "sealing",
    "speed",
    "weight",
    "quantity",
    "volume"
]

process_columns = [
    column for column in batches_df.columns
    if any(
        keyword in column.lower()
        for keyword in process_keywords
    )
]

for column in process_columns:

    if column not in get_numeric_columns(batches_df):
        continue

    plot_numeric_distribution(
        batches_df,
        "production_batches",
        column
    )

    plot_numeric_boxplot(
        batches_df,
        "production_batches",
        column
    )


# ------------------------------------------------------------
# 11. SPECIALIZED QUALITY INSPECTION ANALYSIS
# ------------------------------------------------------------

inspection_df = datasets["quality_inspections"]

print("\n" + "=" * 80)
print("QUALITY INSPECTION ANALYSIS")
print("=" * 80)

inspection_category_columns = [
    column for column in inspection_df.columns
    if any(
        keyword in column.lower()
        for keyword in ["result", "status", "outcome", "inspection_type"]
    )
]

for column in inspection_category_columns:

    counts = (
        inspection_df[column]
        .fillna("Missing")
        .astype(str)
        .value_counts()
    )

    if counts.empty:
        continue

    plt.figure(figsize=(9, 5))

    sns.barplot(
        x=counts.index,
        y=counts.values
    )

    plt.title(f"Inspection Result Analysis - {column}")
    plt.xlabel(column.replace("_", " ").title())
    plt.ylabel("Number of Records")

    plt.xticks(rotation=30)

    save_plot(
        "quality_inspections",
        f"special_{column}"
    )


# ------------------------------------------------------------
# 12. SPECIALIZED INVESTIGATION ANALYSIS
# ------------------------------------------------------------

investigations_df = datasets["investigations"]

print("\n" + "=" * 80)
print("ROOT CAUSE INVESTIGATION ANALYSIS")
print("=" * 80)

root_cause_columns = [
    column for column in investigations_df.columns
    if any(
        keyword in column.lower()
        for keyword in ["root_cause", "cause_category", "investigation_method"]
    )
]

for column in root_cause_columns:

    counts = (
        investigations_df[column]
        .fillna("Missing")
        .astype(str)
        .value_counts()
        .head(15)
    )

    if counts.empty:
        continue

    plt.figure(figsize=(11, 6))

    sns.barplot(
        x=counts.values,
        y=counts.index
    )

    plt.title(f"Investigation Analysis - {column}")
    plt.xlabel("Number of Investigations")
    plt.ylabel(column.replace("_", " ").title())

    save_plot(
        "investigations",
        f"special_{column}"
    )


# ------------------------------------------------------------
# 13. SPECIALIZED CORRECTIVE ACTION ANALYSIS
# ------------------------------------------------------------

actions_df = datasets["corrective_actions"]

print("\n" + "=" * 80)
print("CORRECTIVE ACTION ANALYSIS")
print("=" * 80)

action_columns = [
    column for column in actions_df.columns
    if any(
        keyword in column.lower()
        for keyword in [
            "effectiveness",
            "action_type",
            "status",
            "department"
        ]
    )
]

for column in action_columns:

    counts = (
        actions_df[column]
        .fillna("Missing")
        .astype(str)
        .value_counts()
        .head(15)
    )

    if counts.empty:
        continue

    plt.figure(figsize=(11, 6))

    sns.barplot(
        x=counts.values,
        y=counts.index
    )

    plt.title(f"Corrective Action Analysis - {column}")
    plt.xlabel("Number of Actions")
    plt.ylabel(column.replace("_", " ").title())

    save_plot(
        "corrective_actions",
        f"special_{column}"
    )


# ------------------------------------------------------------
# 14. SAVE EDA SUMMARY REPORTS
# ------------------------------------------------------------

EDA_REPORT_FILE = (
    REPORT_PATH / "individual_dataset_eda_report.xlsx"
)

with pd.ExcelWriter(
    EDA_REPORT_FILE,
    engine="openpyxl"
) as writer:

    dataset_summary_df.to_excel(
        writer,
        sheet_name="Dataset_Summary",
        index=False
    )

    column_summary_df.to_excel(
        writer,
        sheet_name="Column_Summary",
        index=False
    )

print("\nEDA summary report saved:")
print(EDA_REPORT_FILE)


# ------------------------------------------------------------
# 15. FINAL EDA SUMMARY
# ------------------------------------------------------------

total_visualizations = sum(
    len(list(
        (INDIVIDUAL_EDA_PATH / name).glob("*.png")
    ))
    for name in datasets
)

print("\n")
print("=" * 75)
print("ZEROQ AI - INDIVIDUAL DATASET EDA COMPLETED")
print("=" * 75)

print("Total datasets analysed:", len(datasets))

print(
    "Total records analysed:",
    sum(len(df) for df in datasets.values())
)

print("Total columns analysed:", len(column_summary_df))

print("Total visualizations generated:", total_visualizations)

print("\nVisualization directory:")
print(INDIVIDUAL_EDA_PATH)

print("\nEDA report:")
print(EDA_REPORT_FILE)

print("=" * 75)
print("INDIVIDUAL DATASET EDA COMPLETED SUCCESSFULLY")
print("=" * 75)
```
