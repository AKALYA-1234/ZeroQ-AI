```python

# ============================================================
# ZEROQ AI - NOTEBOOK 05
# CROSS-DATASET EXPLORATORY DATA ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
import warnings

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (11, 6)
plt.rcParams["figure.dpi"] = 100

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")


# ------------------------------------------------------------
# 2. DEFINE DATASET FILENAMES
# ------------------------------------------------------------

DATASET_FILES = {
    "products": "products_cleaned.csv",
    "customers": "customers_cleaned.csv",
    "facilities": "facilities_cleaned.csv",
    "production_lines": "production_lines_cleaned.csv",
    "materials": "materials_cleaned.csv",
    "production_batches": "production_batches_cleaned.csv",
    "complaints": "complaints_cleaned.csv",
    "quality_inspections": "quality_inspections_cleaned.csv",
    "investigations": "investigations_cleaned.csv",
    "corrective_actions": "corrective_actions_cleaned.csv",
    "data_dictionary": "data_dictionary_cleaned.csv"
}


# ------------------------------------------------------------
# 3. AUTOMATICALLY LOCATE PROCESSED DATA FOLDER
# ------------------------------------------------------------

print("\nSearching for processed datasets...")

search_roots = [Path.cwd(), *Path.cwd().parents]

PROCESSED_DATA_PATH = None

required_files = set(DATASET_FILES.values())

for root in search_roots:

    for file_path in root.rglob("products_cleaned.csv"):

        candidate_folder = file_path.parent

        available_files = {
            file.name for file in candidate_folder.glob("*.csv")
        }

        if required_files.issubset(available_files):

            PROCESSED_DATA_PATH = candidate_folder
            break

    if PROCESSED_DATA_PATH is not None:
        break

if PROCESSED_DATA_PATH is None:

    raise FileNotFoundError(
        "Could not locate all cleaned datasets. "
        "Execute Notebook 2 first."
    )

PROJECT_PATH = PROCESSED_DATA_PATH.parent

PREPROCESSING_PATH = PROJECT_PATH / "preprocessing"
VISUALIZATION_PATH = PROJECT_PATH / "visualizations"

CROSS_EDA_PATH = VISUALIZATION_PATH / "cross_dataset_eda"

REPORT_PATH = PREPROCESSING_PATH / "reports"

CROSS_EDA_PATH.mkdir(parents=True, exist_ok=True)
REPORT_PATH.mkdir(parents=True, exist_ok=True)

print("\nPROJECT CONFIGURATION")
print("-" * 65)
print("Project:", PROJECT_PATH)
print("Processed data:", PROCESSED_DATA_PATH)
print("Cross-dataset visualizations:", CROSS_EDA_PATH)


# ------------------------------------------------------------
# 4. LOAD ALL CLEANED DATASETS
# ------------------------------------------------------------

datasets = {}

for dataset_name, filename in DATASET_FILES.items():

    file_path = PROCESSED_DATA_PATH / filename

    if not file_path.exists():
        raise FileNotFoundError(file_path)

    datasets[dataset_name] = pd.read_csv(file_path)

    print(
        f"{dataset_name:25} | "
        f"Shape: {datasets[dataset_name].shape}"
    )

print("\nTotal datasets loaded:", len(datasets))


# ------------------------------------------------------------
# 5. CREATE WORKING DATAFRAME REFERENCES
# ------------------------------------------------------------

products = datasets["products"].copy()
customers = datasets["customers"].copy()
facilities = datasets["facilities"].copy()
production_lines = datasets["production_lines"].copy()
materials = datasets["materials"].copy()

batches = datasets["production_batches"].copy()
complaints = datasets["complaints"].copy()
inspections = datasets["quality_inspections"].copy()
investigations = datasets["investigations"].copy()
actions = datasets["corrective_actions"].copy()


# ------------------------------------------------------------
# 6. HELPER FUNCTIONS
# ------------------------------------------------------------

cross_analysis_reports = []


def find_column(df, keywords, exclude=None):

    """
    Find a column using keyword matching.
    Returns the first matching column or None.
    """

    exclude = exclude or []

    for column in df.columns:

        column_lower = column.lower()

        if any(
            keyword in column_lower
            for keyword in keywords
        ) and not any(
            item in column_lower
            for item in exclude
        ):

            return column

    return None


def save_plot(plot_name):

    output_file = CROSS_EDA_PATH / f"{plot_name}.png"

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=150,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print("Saved:", output_file.name)


def plot_grouped_counts(df, group_column, value_column, title, filename):

    if (
        df.empty
        or group_column not in df.columns
        or value_column not in df.columns
    ):
        return

    summary = (
        df.groupby(group_column, dropna=False)[value_column]
        .sum()
        .sort_values(ascending=False)
        .head(15)
    )

    if summary.empty:
        return

    plt.figure(figsize=(11, 6))

    sns.barplot(
        x=summary.values,
        y=summary.index.astype(str)
    )

    plt.title(title)
    plt.xlabel(value_column.replace("_", " ").title())
    plt.ylabel(group_column.replace("_", " ").title())

    save_plot(filename)


def save_analysis_report(df, name):

    cross_analysis_reports.append({
        "Analysis": name,
        "Records": len(df),
        "Columns": len(df.columns)
    })


# ------------------------------------------------------------
# 7. IDENTIFY IMPORTANT COLUMNS
# ------------------------------------------------------------

batch_id = "batch_id"
complaint_id = "complaint_id"
product_id = "product_id"
facility_id = "facility_id"
line_id = "production_line_id"

units_column = find_column(
    batches,
    ["units_produced", "production_quantity", "quantity_produced"]
)

complaint_date_column = find_column(
    complaints,
    ["complaint_date", "reported_date", "received_date", "created_at"]
)

batch_date_column = find_column(
    batches,
    ["production_date", "batch_date", "manufacturing_date"]
)

category_column = find_column(
    complaints,
    ["complaint_category", "category", "complaint_type"],
    exclude=["subcategory"]
)

severity_column = find_column(
    complaints,
    ["severity", "priority"]
)

root_cause_column = find_column(
    investigations,
    ["root_cause", "cause_category", "root_cause_category"]
)

effectiveness_column = find_column(
    actions,
    ["effectiveness", "effectiveness_status"]
)

inspection_result_column = find_column(
    inspections,
    ["inspection_result", "result", "inspection_status", "status"]
)

print("\nIDENTIFIED ANALYSIS COLUMNS")
print("-" * 65)

identified_columns = {
    "Batch ID": batch_id,
    "Complaint ID": complaint_id,
    "Product ID": product_id,
    "Facility ID": facility_id,
    "Production Line ID": line_id,
    "Production Quantity": units_column,
    "Complaint Date": complaint_date_column,
    "Production Date": batch_date_column,
    "Complaint Category": category_column,
    "Severity": severity_column,
    "Root Cause": root_cause_column,
    "Action Effectiveness": effectiveness_column,
    "Inspection Result": inspection_result_column
}

for name, column in identified_columns.items():
    print(f"{name:25}: {column}")


# ------------------------------------------------------------
# 8. PREPARE PRODUCTION BATCH DATA
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PRODUCTION BATCH ANALYSIS")
print("=" * 75)

if batch_id not in batches.columns:
    raise KeyError("batch_id column not found in production_batches.")

if batches[batch_id].isna().any():
    print("Warning: Some production batches have missing batch IDs.")

# Remove invalid batch IDs from analysis copy only.
# Original cleaned datasets remain unchanged.

batches_valid = batches.dropna(
    subset=[batch_id]
).copy()

# Ensure one row per batch before joining.
if batches_valid[batch_id].duplicated().any():

    raise ValueError(
        "Duplicate batch IDs found. Resolve primary key issues "
        "before performing batch-level analysis."
    )

# Convert production quantity to numeric.
if units_column:

    batches_valid[units_column] = pd.to_numeric(
        batches_valid[units_column],
        errors="coerce"
    )

    print("Production quantity column:", units_column)

else:

    print(
        "Production quantity column not identified. "
        "Complaint rates per million units cannot be calculated."
    )


# ------------------------------------------------------------
# 9. AGGREGATE COMPLAINTS AT BATCH LEVEL
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("COMPLAINT ANALYSIS BY PRODUCTION BATCH")
print("=" * 75)

if batch_id not in complaints.columns:

    raise KeyError("batch_id column not found in complaints.")

valid_complaints = complaints.dropna(
    subset=[batch_id]
).copy()

# Count complaints linked to each batch.
batch_complaint_counts = (
    valid_complaints.groupby(batch_id)
    .size()
    .reset_index(name="complaint_count")
)

# Each batch is represented only once.
batch_analysis = batches_valid.merge(
    batch_complaint_counts,
    on=batch_id,
    how="left",
    validate="one_to_one"
)

batch_analysis["complaint_count"] = (
    batch_analysis["complaint_count"]
    .fillna(0)
    .astype(int)
)

batch_analysis["complaint_occurred_analysis"] = (
    batch_analysis["complaint_count"] > 0
).astype(int)

print("Total production batches:", len(batch_analysis))

print(
    "Batches linked to complaints:",
    int(batch_analysis["complaint_occurred_analysis"].sum())
)

print(
    "Batches without linked complaints:",
    int(
        (batch_analysis["complaint_occurred_analysis"] == 0).sum()
    )
)

print("\nBatch-level complaint summary:")

display(
    batch_analysis[
        [
            batch_id,
            "complaint_count",
            "complaint_occurred_analysis"
        ]
    ].head(10)
)


# ------------------------------------------------------------
# 10. PRODUCT-WISE COMPLAINT ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PRODUCT-WISE COMPLAINT ANALYSIS")
print("=" * 75)

product_complaint_analysis = (
    valid_complaints.groupby(product_id)
    .agg(
        total_complaints=(complaint_id, "nunique")
    )
    .reset_index()
)

if product_id in batches_valid.columns:

    product_production = (
        batches_valid.groupby(product_id)
        .agg(
            total_batches=(batch_id, "nunique"),
            total_units=(
                units_column,
                "sum"
            ) if units_column else (batch_id, "nunique")
        )
        .reset_index()
    )

    product_complaint_analysis = (
        product_complaint_analysis.merge(
            product_production,
            on=product_id,
            how="left"
        )
    )

    if units_column:

        product_complaint_analysis[
            "complaints_per_million_units"
        ] = np.where(
            product_complaint_analysis["total_units"] > 0,
            product_complaint_analysis["total_complaints"]
            / product_complaint_analysis["total_units"] * 1_000_000,
            np.nan
        )

display(
    product_complaint_analysis.sort_values(
        "total_complaints",
        ascending=False
    )
)

plot_grouped_counts(
    product_complaint_analysis,
    product_id,
    "total_complaints",
    "Total Complaints by Product",
    "product_wise_complaints"
)

if "complaints_per_million_units" in product_complaint_analysis.columns:

    plot_grouped_counts(
        product_complaint_analysis,
        product_id,
        "complaints_per_million_units",
        "Complaint Rate per Million Units by Product",
        "product_complaint_rate"
    )


# ------------------------------------------------------------
# 11. FACILITY-WISE COMPLAINT ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("FACILITY-WISE COMPLAINT ANALYSIS")
print("=" * 75)

facility_complaint_analysis = pd.DataFrame()

if facility_id in batches_valid.columns:

    facility_production = (
        batches_valid.groupby(facility_id)
        .agg(
            total_batches=(batch_id, "nunique")
        )
        .reset_index()
    )

    if units_column:

        facility_units = (
            batches_valid.groupby(facility_id)[units_column]
            .sum(min_count=1)
            .reset_index(name="total_units")
        )

        facility_production = facility_production.merge(
            facility_units,
            on=facility_id,
            how="left"
        )

    facility_complaints = valid_complaints.merge(
        batches_valid[[batch_id, facility_id]],
        on=batch_id,
        how="inner",
        validate="many_to_one"
    )

    facility_complaint_counts = (
        facility_complaints.groupby(facility_id)
        .agg(
            total_complaints=(complaint_id, "nunique")
        )
        .reset_index()
    )

    facility_complaint_analysis = facility_production.merge(
        facility_complaint_counts,
        on=facility_id,
        how="left"
    )

    facility_complaint_analysis["total_complaints"] = (
        facility_complaint_analysis["total_complaints"]
        .fillna(0)
    )

    if units_column:

        facility_complaint_analysis[
            "complaints_per_million_units"
        ] = np.where(
            facility_complaint_analysis["total_units"] > 0,
            facility_complaint_analysis["total_complaints"]
            / facility_complaint_analysis["total_units"] * 1_000_000,
            np.nan
        )

    display(facility_complaint_analysis)

    plot_grouped_counts(
        facility_complaint_analysis,
        facility_id,
        "total_complaints",
        "Total Complaints by Facility",
        "facility_wise_complaints"
    )

else:

    print("Facility ID not available in production batches.")


# ------------------------------------------------------------
# 12. PRODUCTION LINE-WISE COMPLAINT ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PRODUCTION LINE ANALYSIS")
print("=" * 75)

line_complaint_analysis = pd.DataFrame()

if line_id in batches_valid.columns:

    line_production = (
        batches_valid.groupby(line_id)
        .agg(
            total_batches=(batch_id, "nunique")
        )
        .reset_index()
    )

    if units_column:

        line_units = (
            batches_valid.groupby(line_id)[units_column]
            .sum(min_count=1)
            .reset_index(name="total_units")
        )

        line_production = line_production.merge(
            line_units,
            on=line_id,
            how="left"
        )

    line_complaints = valid_complaints.merge(
        batches_valid[[batch_id, line_id]],
        on=batch_id,
        how="inner",
        validate="many_to_one"
    )

    line_complaint_counts = (
        line_complaints.groupby(line_id)
        .agg(
            total_complaints=(complaint_id, "nunique")
        )
        .reset_index()
    )

    line_complaint_analysis = line_production.merge(
        line_complaint_counts,
        on=line_id,
        how="left"
    )

    line_complaint_analysis["total_complaints"] = (
        line_complaint_analysis["total_complaints"]
        .fillna(0)
    )

    if units_column:

        line_complaint_analysis[
            "complaints_per_million_units"
        ] = np.where(
            line_complaint_analysis["total_units"] > 0,
            line_complaint_analysis["total_complaints"]
            / line_complaint_analysis["total_units"] * 1_000_000,
            np.nan
        )

    display(line_complaint_analysis)

    plot_grouped_counts(
        line_complaint_analysis,
        line_id,
        "total_complaints",
        "Complaints by Production Line",
        "production_line_complaints"
    )

else:

    print("Production line ID unavailable.")


# ------------------------------------------------------------
# 13. COMPLAINT CATEGORY ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("COMPLAINT CATEGORY ANALYSIS")
print("=" * 75)

category_analysis = pd.DataFrame()

if category_column:

    category_analysis = (
        complaints.groupby(category_column, dropna=False)
        .agg(
            total_complaints=(complaint_id, "nunique")
        )
        .reset_index()
        .sort_values(
            "total_complaints",
            ascending=False
        )
    )

    display(category_analysis)

    plot_grouped_counts(
        category_analysis,
        category_column,
        "total_complaints",
        "Complaint Category Distribution",
        "complaint_category_analysis"
    )

else:

    print("Complaint category column not identified.")


# ------------------------------------------------------------
# 14. SEVERITY VS PRODUCT ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("COMPLAINT SEVERITY VS PRODUCT")
print("=" * 75)

severity_product_analysis = pd.DataFrame()

if severity_column and product_id in complaints.columns:

    severity_product_analysis = pd.crosstab(
        complaints[product_id],
        complaints[severity_column],
        dropna=False
    )

    print("Severity distribution by product:")
    display(severity_product_analysis)

    plt.figure(figsize=(12, 7))

    sns.heatmap(
        severity_product_analysis,
        annot=True,
        fmt="g",
        cmap="YlOrRd"
    )

    plt.title("Complaint Severity vs Product")

    plt.xlabel("Severity")
    plt.ylabel("Product ID")

    save_plot("severity_vs_product_heatmap")

else:

    print("Required severity or product columns unavailable.")


# ------------------------------------------------------------
# 15. PRODUCTION PARAMETERS VS COMPLAINT OCCURRENCE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("PRODUCTION PARAMETERS VS COMPLAINT OCCURRENCE")
print("=" * 75)

process_keywords = [
    "temperature",
    "humidity",
    "pressure",
    "sealing",
    "speed",
    "weight"
]

process_columns = [
    column
    for column in batch_analysis.columns
    if any(
        keyword in column.lower()
        for keyword in process_keywords
    )
]

parameter_complaint_reports = []

for column in process_columns:

    values = pd.to_numeric(
        batch_analysis[column],
        errors="coerce"
    )

    analysis_df = batch_analysis[
        [batch_id, "complaint_occurred_analysis"]
    ].copy()

    analysis_df[column] = values

    analysis_df = analysis_df.dropna(
        subset=[column]
    )

    if len(analysis_df) < 10:
        continue

    if analysis_df[column].nunique() < 2:
        continue

    # Divide the parameter into five quantile-based groups.
    try:

        analysis_df["parameter_group"] = pd.qcut(
            analysis_df[column],
            q=5,
            duplicates="drop"
        )

    except ValueError:
        continue

    parameter_summary = (
        analysis_df.groupby(
            "parameter_group",
            observed=False
        )
        .agg(
            total_batches=(batch_id, "nunique"),
            complaint_batches=(
                "complaint_occurred_analysis",
                "sum"
            )
        )
        .reset_index()
    )

    parameter_summary["complaint_percentage"] = (
        parameter_summary["complaint_batches"]
        / parameter_summary["total_batches"] * 100
    )

    parameter_summary["Parameter"] = column

    parameter_complaint_reports.append(
        parameter_summary
    )

    plt.figure(figsize=(11, 6))

    sns.barplot(
        data=parameter_summary,
        x="parameter_group",
        y="complaint_percentage"
    )

    plt.title(
        f"Complaint Percentage Across {column} Ranges"
    )

    plt.xlabel(f"{column} ranges")
    plt.ylabel("Batches with Complaints (%)")

    plt.xticks(rotation=30)

    save_plot(
        f"parameter_{column}_vs_complaints"
    )

if parameter_complaint_reports:

    parameter_complaint_analysis = pd.concat(
        parameter_complaint_reports,
        ignore_index=True
    )

    print("\nParameter vs complaint analysis:")
    display(parameter_complaint_analysis)

else:

    parameter_complaint_analysis = pd.DataFrame()

    print("No suitable numerical process parameters found.")


# ------------------------------------------------------------
# 16. QUALITY INSPECTION VS COMPLAINT ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("QUALITY INSPECTION VS COMPLAINT ANALYSIS")
print("=" * 75)

inspection_batch_analysis = pd.DataFrame()

if (
    batch_id in inspections.columns
    and batch_id in batch_analysis.columns
):

    inspection_temp = inspections.dropna(
        subset=[batch_id]
    ).copy()

    # Aggregate inspection records before joining.
    # This avoids multiplying complaint records when a batch
    # has multiple quality inspections.

    inspection_counts = (
        inspection_temp.groupby(batch_id)
        .size()
        .reset_index(name="inspection_count")
    )

    inspection_batch_analysis = batch_analysis[
        [
            batch_id,
            "complaint_count",
            "complaint_occurred_analysis"
        ]
    ].merge(
        inspection_counts,
        on=batch_id,
        how="left",
        validate="one_to_one"
    )

    inspection_batch_analysis["inspection_count"] = (
        inspection_batch_analysis["inspection_count"]
        .fillna(0)
    )

    print("Inspection count vs complaint occurrence:")
    display(
        inspection_batch_analysis[
            [
                batch_id,
                "inspection_count",
                "complaint_count",
                "complaint_occurred_analysis"
            ]
        ].head(15)
    )

    # Analyse inspection result categories if available.
    if inspection_result_column:

        inspection_temp["inspection_result_normalized"] = (
            inspection_temp[inspection_result_column]
            .fillna("Unknown")
            .astype(str)
            .str.strip()
            .str.lower()
        )

        # Group inspection outcomes by batch.
        inspection_outcome = (
            inspection_temp.groupby(batch_id)
            ["inspection_result_normalized"]
            .agg(
                lambda values: ", ".join(
                    sorted(set(values))
                )
            )
            .reset_index(name="inspection_outcomes")
        )

        inspection_batch_analysis = (
            inspection_batch_analysis.merge(
                inspection_outcome,
                on=batch_id,
                how="left",
                validate="one_to_one"
            )
        )

        outcome_complaint_analysis = (
            inspection_batch_analysis.groupby(
                "inspection_outcomes",
                dropna=False
            )
            .agg(
                total_batches=(batch_id, "nunique"),
                complaint_batches=(
                    "complaint_occurred_analysis",
                    "sum"
                )
            )
            .reset_index()
        )

        outcome_complaint_analysis[
            "complaint_percentage"
        ] = (
            outcome_complaint_analysis["complaint_batches"]
            / outcome_complaint_analysis["total_batches"] * 100
        )

        display(outcome_complaint_analysis)

        plt.figure(figsize=(12, 6))

        sns.barplot(
            data=outcome_complaint_analysis,
            x="inspection_outcomes",
            y="complaint_percentage"
        )

        plt.title(
            "Complaint Percentage by Inspection Outcome"
        )

        plt.xlabel("Inspection Outcomes")
        plt.ylabel("Complaint Percentage (%)")

        plt.xticks(rotation=30)

        save_plot("inspection_outcome_vs_complaints")

else:

    print("Batch relationship unavailable for inspections.")


# ------------------------------------------------------------
# 17. INVESTIGATION ROOT CAUSE VS COMPLAINT CATEGORY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ROOT CAUSE VS COMPLAINT CATEGORY ANALYSIS")
print("=" * 75)

root_cause_category_analysis = pd.DataFrame()

if (
    complaint_id in investigations.columns
    and complaint_id in complaints.columns
    and root_cause_column
    and category_column
):

    investigation_complaint = investigations[
        [complaint_id, root_cause_column]
    ].dropna(
        subset=[complaint_id]
    )

    complaint_category = complaints[
        [complaint_id, category_column]
    ].dropna(
        subset=[complaint_id]
    )

    # Keep the relationship at complaint level.
    # Duplicate investigations are aggregated to avoid
    # inflating complaint counts.

    investigation_complaint = (
        investigation_complaint.groupby(complaint_id)
        [root_cause_column]
        .agg(
            lambda values: ", ".join(
                sorted(set(values.dropna().astype(str)))
            )
        )
        .reset_index()
    )

    root_cause_join = complaint_category.merge(
        investigation_complaint,
        on=complaint_id,
        how="inner",
        validate="one_to_one"
    )

    root_cause_category_analysis = pd.crosstab(
        root_cause_join[category_column],
        root_cause_join[root_cause_column]
    )

    display(root_cause_category_analysis)

    plt.figure(figsize=(14, 8))

    sns.heatmap(
        root_cause_category_analysis,
        annot=True,
        fmt="g",
        cmap="Blues"
    )

    plt.title(
        "Complaint Category vs Investigation Root Cause"
    )

    plt.xlabel("Investigation Root Cause")
    plt.ylabel("Complaint Category")

    save_plot("root_cause_vs_complaint_category")

else:

    print(
        "Required complaint category or investigation root cause "
        "columns are unavailable."
    )


# ------------------------------------------------------------
# 18. CORRECTIVE ACTION EFFECTIVENESS ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CORRECTIVE ACTION EFFECTIVENESS ANALYSIS")
print("=" * 75)

effectiveness_analysis = pd.DataFrame()

if effectiveness_column:

    effectiveness_analysis = (
        actions.groupby(effectiveness_column, dropna=False)
        .size()
        .reset_index(name="total_actions")
        .sort_values(
            "total_actions",
            ascending=False
        )
    )

    display(effectiveness_analysis)

    plot_grouped_counts(
        effectiveness_analysis,
        effectiveness_column,
        "total_actions",
        "Corrective Action Effectiveness Distribution",
        "corrective_action_effectiveness"
    )

else:

    print("Effectiveness column not identified.")


# ------------------------------------------------------------
# 19. CORRECTIVE ACTION VS ROOT CAUSE ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CORRECTIVE ACTION VS ROOT CAUSE")
print("=" * 75)

action_root_cause_analysis = pd.DataFrame()

if (
    "investigation_id" in actions.columns
    and "investigation_id" in investigations.columns
    and root_cause_column
    and effectiveness_column
):

    investigation_root = investigations[
        ["investigation_id", root_cause_column]
    ].dropna(
        subset=["investigation_id"]
    )

    # Aggregate root causes by investigation.
    investigation_root = (
        investigation_root.groupby("investigation_id")
        [root_cause_column]
        .agg(
            lambda values: ", ".join(
                sorted(set(values.dropna().astype(str)))
            )
        )
        .reset_index()
    )

    action_root = actions[
        ["investigation_id", effectiveness_column]
    ].dropna(
        subset=["investigation_id"]
    )

    action_root_join = action_root.merge(
        investigation_root,
        on="investigation_id",
        how="inner"
    )

    action_root_cause_analysis = pd.crosstab(
        action_root_join[root_cause_column],
        action_root_join[effectiveness_column]
    )

    display(action_root_cause_analysis)

    plt.figure(figsize=(13, 7))

    sns.heatmap(
        action_root_cause_analysis,
        annot=True,
        fmt="g",
        cmap="YlGnBu"
    )

    plt.title(
        "Root Cause vs Corrective Action Effectiveness"
    )

    plt.xlabel("Action Effectiveness")
    plt.ylabel("Root Cause")

    save_plot("root_cause_vs_action_effectiveness")

else:

    print("Required investigation/action columns unavailable.")


# ------------------------------------------------------------
# 20. RECURRING COMPLAINT ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("RECURRING COMPLAINT ANALYSIS")
print("=" * 75)

recurring_complaint_analysis = pd.DataFrame()

if category_column and product_id in complaints.columns:

    recurring_complaint_analysis = (
        complaints.groupby(
            [product_id, category_column],
            dropna=False
        )
        .agg(
            total_complaints=(complaint_id, "nunique")
        )
        .reset_index()
        .sort_values(
            "total_complaints",
            ascending=False
        )
    )

    display(
        recurring_complaint_analysis.head(30)
    )

    recurring_plot = recurring_complaint_analysis.head(15)

    if not recurring_plot.empty:

        plt.figure(figsize=(12, 7))

        sns.barplot(
            data=recurring_plot,
            x="total_complaints",
            y=product_id,
            hue=category_column
        )

        plt.title(
            "Recurring Complaint Categories by Product"
        )

        plt.xlabel("Number of Complaints")
        plt.ylabel("Product ID")

        save_plot("recurring_complaints_by_product")

else:

    print("Required product or complaint category columns unavailable.")


# ------------------------------------------------------------
# 21. YEARLY COMPLAINT TREND
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("YEARLY COMPLAINT TREND")
print("=" * 75)

yearly_complaint_analysis = pd.DataFrame()

if complaint_date_column:

    complaints_date_temp = complaints.copy()

    complaints_date_temp[complaint_date_column] = (
        pd.to_datetime(
            complaints_date_temp[complaint_date_column],
            errors="coerce"
        )
    )

    complaints_date_temp = complaints_date_temp.dropna(
        subset=[complaint_date_column]
    )

    yearly_complaint_analysis = (
        complaints_date_temp
        .groupby(
            complaints_date_temp[complaint_date_column].dt.year
        )
        .agg(
            total_complaints=(complaint_id, "nunique")
        )
        .reset_index()
    )

    yearly_complaint_analysis.columns = [
        "Year",
        "Total_Complaints"
    ]

    display(yearly_complaint_analysis)

    if len(yearly_complaint_analysis) > 1:

        plt.figure(figsize=(13, 6))

        sns.lineplot(
            data=yearly_complaint_analysis,
            x="Year",
            y="Total_Complaints",
            marker="o"
        )

        plt.title("Yearly Customer Complaint Trend")
        plt.xlabel("Year")
        plt.ylabel("Number of Complaints")

        plt.xticks(rotation=45)

        save_plot("yearly_complaint_trend")

else:

    print("Complaint date column unavailable.")


# ------------------------------------------------------------
# 22. SAVE CROSS-DATASET ANALYSIS TABLES
# ------------------------------------------------------------

CROSS_REPORT_FILE = (
    REPORT_PATH / "cross_dataset_eda_report.xlsx"
)

with pd.ExcelWriter(
    CROSS_REPORT_FILE,
    engine="openpyxl"
) as writer:

    dataset_summary = pd.DataFrame([
        {
            "Dataset": name,
            "Rows": len(df),
            "Columns": len(df.columns)
        }
        for name, df in datasets.items()
    ])

    dataset_summary.to_excel(
        writer,
        sheet_name="Dataset_Summary",
        index=False
    )

    batch_analysis.to_excel(
        writer,
        sheet_name="Batch_Complaints",
        index=False
    )

    product_complaint_analysis.to_excel(
        writer,
        sheet_name="Product_Complaints",
        index=False
    )

    if not facility_complaint_analysis.empty:
        facility_complaint_analysis.to_excel(
            writer,
            sheet_name="Facility_Complaints",
            index=False
        )

    if not line_complaint_analysis.empty:
        line_complaint_analysis.to_excel(
            writer,
            sheet_name="Line_Complaints",
            index=False
        )

    category_analysis.to_excel(
        writer,
        sheet_name="Complaint_Categories",
        index=False
    )

    severity_product_analysis.to_excel(
        writer,
        sheet_name="Severity_Product",
        index=True
    )

    parameter_complaint_analysis.to_excel(
        writer,
        sheet_name="Process_Parameters",
        index=False
    )

    inspection_batch_analysis.to_excel(
        writer,
        sheet_name="Inspection_Complaints",
        index=False
    )

    root_cause_category_analysis.to_excel(
        writer,
        sheet_name="Root_Cause_Category",
        index=True
    )

    effectiveness_analysis.to_excel(
        writer,
        sheet_name="Action_Effectiveness",
        index=False
    )

    action_root_cause_analysis.to_excel(
        writer,
        sheet_name="Action_Root_Cause",
        index=True
    )

    recurring_complaint_analysis.to_excel(
        writer,
        sheet_name="Recurring_Complaints",
        index=False
    )

    yearly_complaint_analysis.to_excel(
        writer,
        sheet_name="Yearly_Complaints",
        index=False
    )

print("\nCross-dataset report saved:")
print(CROSS_REPORT_FILE)


# ------------------------------------------------------------
# 23. FINAL SUMMARY
# ------------------------------------------------------------

total_visualizations = len(
    list(CROSS_EDA_PATH.glob("*.png"))
)

print("\n")
print("=" * 75)
print("ZEROQ AI - CROSS-DATASET EDA COMPLETED")
print("=" * 75)

print("Total datasets analysed:", len(datasets))

print("Total production batches:", len(batches_valid))

print("Total complaints:", len(complaints))

print(
    "Batches linked to complaints:",
    int(batch_analysis["complaint_occurred_analysis"].sum())
)

print("Total cross-dataset visualizations:", total_visualizations)

print("\nVisualization directory:")
print(CROSS_EDA_PATH)

print("\nAnalysis report:")
print(CROSS_REPORT_FILE)

print("=" * 75)
print("CROSS-DATASET EDA COMPLETED")
print("=" * 75)
```
