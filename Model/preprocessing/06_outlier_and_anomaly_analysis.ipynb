```python

# ============================================================
# ZEROQ AI - NOTEBOOK 06
# OUTLIER AND ANOMALY ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from sklearn.ensemble import IsolationForest
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

import warnings

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (11, 6)
plt.rcParams["figure.dpi"] = 100

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")


# ------------------------------------------------------------
# 2. DEFINE DATASET FILENAMES
# ------------------------------------------------------------

DATASET_FILES = {
    "products": "products_cleaned.csv",
    "customers": "customers_cleaned.csv",
    "facilities": "facilities_cleaned.csv",
    "production_lines": "production_lines_cleaned.csv",
    "materials": "materials_cleaned.csv",
    "production_batches": "production_batches_cleaned.csv",
    "complaints": "complaints_cleaned.csv",
    "quality_inspections": "quality_inspections_cleaned.csv",
    "investigations": "investigations_cleaned.csv",
    "corrective_actions": "corrective_actions_cleaned.csv",
    "data_dictionary": "data_dictionary_cleaned.csv"
}

REQUIRED_FILES = set(DATASET_FILES.values())


# ------------------------------------------------------------
# 3. AUTOMATICALLY LOCATE PROCESSED DATA FOLDER
# ------------------------------------------------------------

print("\nSearching for processed datasets...")

search_roots = [Path.cwd(), *Path.cwd().parents]

PROCESSED_DATA_PATH = None

for root in search_roots:

    for file_path in root.rglob("products_cleaned.csv"):

        candidate_folder = file_path.parent

        available_files = {
            file.name for file in candidate_folder.glob("*.csv")
        }

        if REQUIRED_FILES.issubset(available_files):

            PROCESSED_DATA_PATH = candidate_folder
            break

    if PROCESSED_DATA_PATH is not None:
        break

if PROCESSED_DATA_PATH is None:

    raise FileNotFoundError(
        "Could not locate all 11 cleaned datasets. "
        "Execute Notebook 2 first."
    )

PROJECT_PATH = PROCESSED_DATA_PATH.parent

PREPROCESSING_PATH = PROJECT_PATH / "preprocessing"
VISUALIZATION_PATH = PROJECT_PATH / "visualizations"

ANOMALY_PATH = VISUALIZATION_PATH / "outlier_anomaly_analysis"

REPORT_PATH = PREPROCESSING_PATH / "reports"

ANOMALY_PATH.mkdir(parents=True, exist_ok=True)
REPORT_PATH.mkdir(parents=True, exist_ok=True)

print("\nPROJECT CONFIGURATION")
print("-" * 65)
print("Project:", PROJECT_PATH)
print("Processed data:", PROCESSED_DATA_PATH)
print("Anomaly visualizations:", ANOMALY_PATH)


# ------------------------------------------------------------
# 4. LOAD ALL CLEANED DATASETS
# ------------------------------------------------------------

datasets = {}

for dataset_name, filename in DATASET_FILES.items():

    file_path = PROCESSED_DATA_PATH / filename

    if not file_path.exists():
        raise FileNotFoundError(file_path)

    datasets[dataset_name] = pd.read_csv(file_path)

    print(
        f"{dataset_name:25} | "
        f"Shape: {datasets[dataset_name].shape}"
    )

print("\nTotal datasets loaded:", len(datasets))


# ------------------------------------------------------------
# 5. SELECT PRODUCTION BATCH DATA
# ------------------------------------------------------------

batches = datasets["production_batches"].copy()
complaints = datasets["complaints"].copy()
inspections = datasets["quality_inspections"].copy()

if "batch_id" not in batches.columns:
    raise KeyError("batch_id not found in production_batches.")

if batches["batch_id"].duplicated().any():
    raise ValueError(
        "Duplicate batch IDs found. Resolve primary key issues first."
    )

print("\nTotal production batches:", len(batches))


# ------------------------------------------------------------
# 6. IDENTIFY NUMERICAL PROCESS PARAMETERS
# ------------------------------------------------------------

# Exclude IDs, date fields and obvious record flags from
# manufacturing measurement analysis.

excluded_keywords = [
    "_id",
    "date",
    "timestamp",
    "year",
    "month",
    "day",
    "complaint_occurred",
    "complaint_count",
    "linked",
    "flag",
    "is_"
]

numeric_columns = batches.select_dtypes(
    include=np.number
).columns.tolist()

process_columns = [
    column
    for column in numeric_columns
    if not any(
        keyword in column.lower()
        for keyword in excluded_keywords
    )
    and batches[column].nunique(dropna=True) > 1
]

print("\nIDENTIFIED NUMERICAL PROCESS PARAMETERS")
print("-" * 65)

for column in process_columns:
    print(column)

print("\nTotal process parameters identified:", len(process_columns))

if not process_columns:
    raise ValueError(
        "No suitable numerical process parameters found. "
        "Inspect the production_batches columns."
    )


# ------------------------------------------------------------
# 7. BASIC NUMERICAL DATA QUALITY CHECK
# ------------------------------------------------------------

data_quality_reports = []

for column in process_columns:

    values = pd.to_numeric(
        batches[column],
        errors="coerce"
    )

    data_quality_reports.append({

        "Column": column,

        "Total_Records": len(values),

        "Missing_Values": int(values.isna().sum()),

        "Infinite_Values": int(
            np.isinf(values.dropna()).sum()
        ),

        "Unique_Values": int(values.nunique(dropna=True)),

        "Minimum": values.min(),

        "Maximum": values.max(),

        "Mean": values.mean(),

        "Median": values.median(),

        "Standard_Deviation": values.std()

    })


data_quality_report_df = pd.DataFrame(
    data_quality_reports
)

print("\nNUMERICAL DATA QUALITY REPORT")
display(data_quality_report_df)


# ------------------------------------------------------------
# 8. IQR OUTLIER DETECTION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("IQR OUTLIER DETECTION")
print("=" * 75)

iqr_reports = []
iqr_outlier_records = []

for column in process_columns:

    values = pd.to_numeric(
        batches[column],
        errors="coerce"
    )

    valid_values = values.replace(
        [np.inf, -np.inf],
        np.nan
    ).dropna()

    if len(valid_values) < 4:
        continue

    Q1 = valid_values.quantile(0.25)
    Q3 = valid_values.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_mask = (
        (values < lower_bound)
        | (values > upper_bound)
    )

    outlier_count = int(outlier_mask.sum())

    outlier_percentage = (
        outlier_count / len(valid_values) * 100
    )

    iqr_reports.append({

        "Parameter": column,

        "Q1": Q1,

        "Q3": Q3,

        "IQR": IQR,

        "Lower_Bound": lower_bound,

        "Upper_Bound": upper_bound,

        "Outlier_Count": outlier_count,

        "Valid_Records": len(valid_values),

        "Outlier_Percentage": round(
            outlier_percentage, 2
        )

    })

    if outlier_count > 0:

        temp = batches.loc[
            outlier_mask,
            ["batch_id", column]
        ].copy()

        temp["Parameter"] = column

        temp["Detection_Method"] = "IQR"

        temp["Lower_Bound"] = lower_bound

        temp["Upper_Bound"] = upper_bound

        iqr_outlier_records.append(temp)


iqr_report_df = pd.DataFrame(iqr_reports)

if iqr_outlier_records:

    iqr_outliers_df = pd.concat(
        iqr_outlier_records,
        ignore_index=True
    )

else:

    iqr_outliers_df = pd.DataFrame(
        columns=[
            "batch_id",
            "Parameter",
            "Detection_Method",
            "Lower_Bound",
            "Upper_Bound"
        ]
    )

display(iqr_report_df)

print("\nTotal IQR outlier records:", len(iqr_outliers_df))


# ------------------------------------------------------------
# 9. VISUALIZE IQR OUTLIERS USING BOXPLOTS
# ------------------------------------------------------------

for column in process_columns:

    values = pd.to_numeric(
        batches[column],
        errors="coerce"
    ).replace(
        [np.inf, -np.inf],
        np.nan
    ).dropna()

    if len(values) < 2:
        continue

    plt.figure(figsize=(10, 5))

    sns.boxplot(x=values)

    plt.title(
        f"IQR Outlier Analysis - {column}"
    )

    plt.xlabel(column.replace("_", " ").title())

    output_file = ANOMALY_PATH / f"iqr_boxplot_{column}.png"

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=150,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# 10. Z-SCORE OUTLIER DETECTION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("Z-SCORE OUTLIER DETECTION")
print("=" * 75)

zscore_reports = []
zscore_outlier_records = []

for column in process_columns:

    values = pd.to_numeric(
        batches[column],
        errors="coerce"
    )

    values = values.replace(
        [np.inf, -np.inf],
        np.nan
    )

    mean_value = values.mean()
    std_value = values.std()

    if pd.isna(std_value) or std_value == 0:
        continue

    z_scores = (values - mean_value) / std_value

    outlier_mask = z_scores.abs() > 3

    outlier_count = int(outlier_mask.sum())

    valid_count = int(values.notna().sum())

    zscore_reports.append({

        "Parameter": column,

        "Mean": mean_value,

        "Standard_Deviation": std_value,

        "Outlier_Count": outlier_count,

        "Valid_Records": valid_count,

        "Outlier_Percentage": round(
            outlier_count / valid_count * 100, 2
        ) if valid_count > 0 else 0

    })

    if outlier_count > 0:

        temp = batches.loc[
            outlier_mask,
            ["batch_id", column]
        ].copy()

        temp["Parameter"] = column

        temp["Z_Score"] = z_scores.loc[outlier_mask].values

        temp["Detection_Method"] = "Z-Score"

        zscore_outlier_records.append(temp)


zscore_report_df = pd.DataFrame(zscore_reports)

if zscore_outlier_records:

    zscore_outliers_df = pd.concat(
        zscore_outlier_records,
        ignore_index=True
    )

else:

    zscore_outliers_df = pd.DataFrame(
        columns=[
            "batch_id",
            "Parameter",
            "Z_Score",
            "Detection_Method"
        ]
    )

display(zscore_report_df)


# ------------------------------------------------------------
# 11. MODIFIED Z-SCORE USING MEDIAN ABSOLUTE DEVIATION
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MODIFIED Z-SCORE ANALYSIS")
print("=" * 75)

modified_z_reports = []
modified_z_outlier_records = []

for column in process_columns:

    values = pd.to_numeric(
        batches[column],
        errors="coerce"
    ).replace(
        [np.inf, -np.inf],
        np.nan
    )

    median_value = values.median()

    MAD = np.median(
        np.abs(values.dropna() - median_value)
    )

    if pd.isna(MAD) or MAD == 0:
        continue

    modified_z_scores = (
        0.6745 * (values - median_value) / MAD
    )

    outlier_mask = modified_z_scores.abs() > 3.5

    outlier_count = int(outlier_mask.sum())

    valid_count = int(values.notna().sum())

    modified_z_reports.append({

        "Parameter": column,

        "Median": median_value,

        "MAD": MAD,

        "Outlier_Count": outlier_count,

        "Valid_Records": valid_count,

        "Outlier_Percentage": round(
            outlier_count / valid_count * 100, 2
        ) if valid_count > 0 else 0

    })

    if outlier_count > 0:

        temp = batches.loc[
            outlier_mask,
            ["batch_id", column]
        ].copy()

        temp["Parameter"] = column

        temp["Modified_Z_Score"] = (
            modified_z_scores.loc[outlier_mask].values
        )

        temp["Detection_Method"] = "Modified Z-Score"

        modified_z_outlier_records.append(temp)


modified_z_report_df = pd.DataFrame(modified_z_reports)

if modified_z_outlier_records:

    modified_z_outliers_df = pd.concat(
        modified_z_outlier_records,
        ignore_index=True
    )

else:

    modified_z_outliers_df = pd.DataFrame(
        columns=[
            "batch_id",
            "Parameter",
            "Modified_Z_Score",
            "Detection_Method"
        ]
    )

display(modified_z_report_df)


# ------------------------------------------------------------
# 12. MULTIVARIATE ANOMALY DETECTION - ISOLATION FOREST
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ISOLATION FOREST ANOMALY DETECTION")
print("=" * 75)

isolation_forest_results = batches[
    ["batch_id"]
].copy()

# Exclude rows where every selected parameter is missing
X = batches[process_columns].copy()

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

valid_row_mask = X.notna().any(axis=1)

X_valid = X.loc[valid_row_mask].copy()

isolation_forest_results["IF_Anomaly_Score"] = np.nan

isolation_forest_results["IF_Anomaly_Flag"] = np.nan

if len(X_valid) >= 10:

    # Remove columns with no observed values
    X_valid = X_valid.dropna(
        axis=1,
        how="all"
    )

    # Remove constant columns
    X_valid = X_valid.loc[
        :,
        X_valid.nunique(dropna=True) > 1
    ]

    if X_valid.shape[1] > 0:

        # Impute missing measurements using column medians.
        # This transformation is only for anomaly detection.
        imputer = SimpleImputer(strategy="median")

        X_imputed = imputer.fit_transform(X_valid)

        # Standardize measurement scales.
        scaler = StandardScaler()

        X_scaled = scaler.fit_transform(X_imputed)

        isolation_model = IsolationForest(
            n_estimators=200,
            contamination="auto",
            random_state=42,
            n_jobs=-1
        )

        predictions = isolation_model.fit_predict(
            X_scaled
        )

        anomaly_scores = isolation_model.decision_function(
            X_scaled
        )

        # Isolation Forest returns:
        # -1 = Anomaly
        #  1 = Normal

        isolation_forest_results.loc[
            X_valid.index,
            "IF_Anomaly_Flag"
        ] = (
            predictions == -1
        ).astype(int)

        isolation_forest_results.loc[
            X_valid.index,
            "IF_Anomaly_Score"
        ] = anomaly_scores

        isolation_forest_results[
            "IF_Anomaly_Flag"
        ] = isolation_forest_results[
            "IF_Anomaly_Flag"
        ].fillna(0).astype(int)

        print("Isolation Forest completed.")

    else:

        print("No variable numerical columns available.")

else:

    print("Insufficient valid records for Isolation Forest.")


display(
    isolation_forest_results.head(15)
)


# ------------------------------------------------------------
# 13. COMBINE ANOMALY DETECTION RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("COMBINED ANOMALY ANALYSIS")
print("=" * 75)

batch_anomaly_analysis = batches[
    ["batch_id"]
].copy()

# Add Isolation Forest result
batch_anomaly_analysis = batch_anomaly_analysis.merge(
    isolation_forest_results,
    on="batch_id",
    how="left",
    validate="one_to_one"
)

# Number of IQR-detected parameter outliers per batch
if not iqr_outliers_df.empty:

    iqr_batch_counts = (
        iqr_outliers_df.groupby("batch_id")
        .size()
        .reset_index(name="IQR_Outlier_Count")
    )

else:

    iqr_batch_counts = pd.DataFrame(
        columns=["batch_id", "IQR_Outlier_Count"]
    )

batch_anomaly_analysis = batch_anomaly_analysis.merge(
    iqr_batch_counts,
    on="batch_id",
    how="left",
    validate="one_to_one"
)

# Number of Z-score-detected parameter outliers per batch
if not zscore_outliers_df.empty:

    zscore_batch_counts = (
        zscore_outliers_df.groupby("batch_id")
        .size()
        .reset_index(name="ZScore_Outlier_Count")
    )

else:

    zscore_batch_counts = pd.DataFrame(
        columns=["batch_id", "ZScore_Outlier_Count"]
    )

batch_anomaly_analysis = batch_anomaly_analysis.merge(
    zscore_batch_counts,
    on="batch_id",
    how="left",
    validate="one_to_one"
)

# Number of Modified Z-score outliers per batch
if not modified_z_outliers_df.empty:

    modified_batch_counts = (
        modified_z_outliers_df.groupby("batch_id")
        .size()
        .reset_index(name="Modified_Z_Outlier_Count")
    )

else:

    modified_batch_counts = pd.DataFrame(
        columns=["batch_id", "Modified_Z_Outlier_Count"]
    )

batch_anomaly_analysis = batch_anomaly_analysis.merge(
    modified_batch_counts,
    on="batch_id",
    how="left",
    validate="one_to_one"
)

count_columns = [
    "IQR_Outlier_Count",
    "ZScore_Outlier_Count",
    "Modified_Z_Outlier_Count"
]

for column in count_columns:

    batch_anomaly_analysis[column] = (
        batch_anomaly_analysis[column]
        .fillna(0)
        .astype(int)
    )

# Any statistical method detects at least one anomaly.
batch_anomaly_analysis["Statistical_Anomaly_Flag"] = (
    batch_anomaly_analysis[count_columns].sum(axis=1) > 0
).astype(int)

print("\nCombined batch anomaly results:")

display(
    batch_anomaly_analysis.head(20)
)


# ------------------------------------------------------------
# 14. ANOMALY VS CUSTOMER COMPLAINT ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ANOMALY VS CUSTOMER COMPLAINT ANALYSIS")
print("=" * 75)

if "batch_id" not in complaints.columns:

    print("Complaint batch ID unavailable.")

else:

    valid_complaints = complaints.dropna(
        subset=["batch_id"]
    )

    complaint_batch_counts = (
        valid_complaints.groupby("batch_id")
        .size()
        .reset_index(name="complaint_count")
    )

    anomaly_complaint_analysis = batch_anomaly_analysis.merge(
        complaint_batch_counts,
        on="batch_id",
        how="left",
        validate="one_to_one"
    )

    anomaly_complaint_analysis["complaint_count"] = (
        anomaly_complaint_analysis["complaint_count"]
        .fillna(0)
    )

    anomaly_complaint_analysis["complaint_occurred"] = (
        anomaly_complaint_analysis["complaint_count"] > 0
    ).astype(int)

    anomaly_complaint_summary = (
        anomaly_complaint_analysis.groupby(
            "IF_Anomaly_Flag"
        )
        .agg(
            total_batches=("batch_id", "nunique"),
            complaint_batches=("complaint_occurred", "sum"),
            total_complaints=("complaint_count", "sum")
        )
        .reset_index()
    )

    anomaly_complaint_summary["complaint_percentage"] = (
        anomaly_complaint_summary["complaint_batches"]
        / anomaly_complaint_summary["total_batches"] * 100
    )

    anomaly_complaint_summary["Anomaly_Status"] = (
        anomaly_complaint_summary["IF_Anomaly_Flag"]
        .map({
            0: "Normal",
            1: "Anomaly"
        })
    )

    display(anomaly_complaint_summary)

    plt.figure(figsize=(9, 6))

    sns.barplot(
        data=anomaly_complaint_summary,
        x="Anomaly_Status",
        y="complaint_percentage"
    )

    plt.title(
        "Complaint Percentage: Normal vs Anomalous Batches"
    )

    plt.xlabel("Batch Classification")
    plt.ylabel("Complaint Percentage (%)")

    output_file = (
        ANOMALY_PATH / "anomaly_vs_complaint_rate.png"
    )

    plt.tight_layout()
    plt.savefig(output_file, dpi=150, bbox_inches="tight")
    plt.show()
    plt.close()

    print("Saved:", output_file.name)


# ------------------------------------------------------------
# 15. ANOMALY VS QUALITY INSPECTION ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("ANOMALY VS QUALITY INSPECTION ANALYSIS")
print("=" * 75)

if (
    "batch_id" in inspections.columns
    and len(inspections) > 0
):

    inspection_temp = inspections.dropna(
        subset=["batch_id"]
    ).copy()

    inspection_result_column = next(
        (
            column for column in inspection_temp.columns
            if any(
                keyword in column.lower()
                for keyword in ["inspection_result", "result", "inspection_status"]
            )
        ),
        None
    )

    if inspection_result_column:

        # Aggregate inspection outcomes at batch level.
        inspection_temp["inspection_result_normalized"] = (
            inspection_temp[inspection_result_column]
            .fillna("Unknown")
            .astype(str)
            .str.strip()
            .str.lower()
        )

        inspection_summary = (
            inspection_temp.groupby("batch_id")
            ["inspection_result_normalized"]
            .agg(
                lambda values: ", ".join(
                    sorted(set(values))
                )
            )
            .reset_index(name="inspection_outcomes")
        )

        inspection_anomaly_analysis = (
            batch_anomaly_analysis.merge(
                inspection_summary,
                on="batch_id",
                how="left",
                validate="one_to_one"
            )
        )

        inspection_anomaly_analysis["inspection_outcomes"] = (
            inspection_anomaly_analysis[
                "inspection_outcomes"
            ].fillna("No Inspection Record")
        )

        display(
            inspection_anomaly_analysis[
                [
                    "batch_id",
                    "IF_Anomaly_Flag",
                    "inspection_outcomes"
                ]
            ].head(20)
        )

    else:

        inspection_anomaly_analysis = pd.DataFrame()

        print("Inspection result column not found.")

else:

    inspection_anomaly_analysis = pd.DataFrame()

    print("Inspection batch relationship unavailable.")


# ------------------------------------------------------------
# 16. IDENTIFY EXTREME BATCHES FOR REVIEW
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("EXTREME BATCH IDENTIFICATION")
print("=" * 75)

extreme_batches = batches[
    ["batch_id"] + process_columns
].copy()

extreme_batches = extreme_batches.merge(
    batch_anomaly_analysis[
        [
            "batch_id",
            "Statistical_Anomaly_Flag",
            "IF_Anomaly_Flag"
        ]
    ],
    on="batch_id",
    how="left",
    validate="one_to_one"
)

extreme_batches["Review_Required"] = (
    (
        extreme_batches["Statistical_Anomaly_Flag"] == 1
    )
    |
    (
        extreme_batches["IF_Anomaly_Flag"] == 1
    )
).astype(int)

extreme_batches = extreme_batches[
    extreme_batches["Review_Required"] == 1
].copy()

print("Batches flagged for further review:", len(extreme_batches))

display(extreme_batches.head(30))


# ------------------------------------------------------------
# 17. CREATE OUTLIER AND ANOMALY SUMMARY
# ------------------------------------------------------------

anomaly_summary = pd.DataFrame({

    "Detection_Method": [
        "IQR",
        "Z-Score",
        "Modified Z-Score",
        "Isolation Forest"
    ],

    "Detected_Records": [

        len(iqr_outliers_df),

        len(zscore_outliers_df),

        len(modified_z_outliers_df),

        int(
            batch_anomaly_analysis["IF_Anomaly_Flag"].sum()
        )

    ]

})

print("\nANOMALY DETECTION SUMMARY")
display(anomaly_summary)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=anomaly_summary,
    x="Detection_Method",
    y="Detected_Records"
)

plt.title("Comparison of Anomaly Detection Methods")

plt.xlabel("Detection Method")
plt.ylabel("Number of Detected Records")

plt.xticks(rotation=20)

output_file = ANOMALY_PATH / "anomaly_method_comparison.png"

plt.tight_layout()
plt.savefig(output_file, dpi=150, bbox_inches="tight")
plt.show()
plt.close()


# ------------------------------------------------------------
# 18. SAVE ALL ANALYSIS REPORTS
# ------------------------------------------------------------

ANOMALY_REPORT_FILE = (
    REPORT_PATH / "outlier_anomaly_analysis_report.xlsx"
)

with pd.ExcelWriter(
    ANOMALY_REPORT_FILE,
    engine="openpyxl"
) as writer:

    data_quality_report_df.to_excel(
        writer,
        sheet_name="Data_Quality",
        index=False
    )

    iqr_report_df.to_excel(
        writer,
        sheet_name="IQR_Summary",
        index=False
    )

    iqr_outliers_df.to_excel(
        writer,
        sheet_name="IQR_Outliers",
        index=False
    )

    zscore_report_df.to_excel(
        writer,
        sheet_name="ZScore_Summary",
        index=False
    )

    zscore_outliers_df.to_excel(
        writer,
        sheet_name="ZScore_Outliers",
        index=False
    )

    modified_z_report_df.to_excel(
        writer,
        sheet_name="Modified_Z_Summary",
        index=False
    )

    modified_z_outliers_df.to_excel(
        writer,
        sheet_name="Modified_Z_Outliers",
        index=False
    )

    batch_anomaly_analysis.to_excel(
        writer,
        sheet_name="Batch_Anomalies",
        index=False
    )

    anomaly_summary.to_excel(
        writer,
        sheet_name="Anomaly_Comparison",
        index=False
    )

    extreme_batches.to_excel(
        writer,
        sheet_name="Extreme_Batches",
        index=False
    )

    if "anomaly_complaint_summary" in globals():

        anomaly_complaint_summary.to_excel(
            writer,
            sheet_name="Anomaly_Complaints",
            index=False
        )

    if not inspection_anomaly_analysis.empty:

        inspection_anomaly_analysis.to_excel(
            writer,
            sheet_name="Anomaly_Inspections",
            index=False
        )

print("\nAnomaly analysis report saved:")
print(ANOMALY_REPORT_FILE)


# ------------------------------------------------------------
# 19. FINAL SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("ZEROQ AI - OUTLIER AND ANOMALY ANALYSIS COMPLETED")
print("=" * 75)

print("Total production batches:", len(batches))

print("Process parameters analysed:", len(process_columns))

print("IQR outlier records:", len(iqr_outliers_df))

print("Z-score outlier records:", len(zscore_outliers_df))

print("Modified Z-score outlier records:", len(modified_z_outliers_df))

print(
    "Isolation Forest anomalies:",
    int(batch_anomaly_analysis["IF_Anomaly_Flag"].sum())
)

print("Batches requiring review:", len(extreme_batches))

print("\nVisualization directory:")
print(ANOMALY_PATH)

print("\nAnalysis report:")
print(ANOMALY_REPORT_FILE)

print("\nIMPORTANT:")
print("No outliers or anomalous batches were deleted.")
print("All flagged records are retained for investigation.")

print("=" * 75)
print("OUTLIER AND ANOMALY ANALYSIS COMPLETED")
print("=" * 75)
```
